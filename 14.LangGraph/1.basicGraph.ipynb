```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class MsgState(TypedDict):
    message: str
```

```python
def greetings_node(state: MsgState) -> MsgState:
    """Simple Node that's add a greetings MSG to the state"""
    state['message'] = "Hello " + state['message'] + ". wellcome to langgraph"
    return state
```

```python
graph = StateGraph(MsgState)
graph.add_node("initial", greetings_node)

graph.add_edge(START, "initial")
graph.add_edge("initial", END)
bot = graph.compile()
bot
```

```python
response = bot.invoke({"message": "Learning Hub"})
response
```
